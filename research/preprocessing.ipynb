### Reading files in sub-folders in "CSI dataset for wireless human sensing/dataset" and transforming the format of the read data to one which can be used to train a Human Activity Recognition model

```python
import pandas as pd
import numpy as np
import os
import scipy.io as sio
from datetime import datetime
```

```python
# Read all files in subfolders in "CSI dataset for wireless human sensing/dataset"

# Path to the dataset
path = 'CSI dataset for wireless human sensing\\dataset' # edit path as per your system

# Get all subfolders
subfolders = [f.path for f in os.scandir(path) if f.is_dir()]

files = {}

for subfolder in subfolders:
    # Get all files in subfolder
    files[subfolder.split('\\').pop()] = [f.path for f in os.scandir(subfolder) if f.is_file()]

files
```

```python
activity_map = {
    'W': 'walking',
    'R': 'running',
    'J': 'jumping',
    'L': 'sitting still',
    'S': 'standing still',
    'C': 'sitting down/standing up',
    'G': 'arm exercises',
    'E': 'empty',
    'H': 'unknown'
}
```

```python
# df = pd.DataFrame()
activity_data = {}
```

```python
# iterate through every key and value pair in files
for folder, filepaths in files.items():
    # iterate through every file in the folder

    print("Now working on folder: " + folder)
    for filepath in filepaths:
        print("Now working on file: " + filepath)

        mat_contents = sio.loadmat(filepath)
        csi_data = pd.DataFrame.from_dict(mat_contents['csi_buff'])

        print("CSI data from file loaded")

        # get the activity from the filename
        filename = filepath.split('\\').pop()
        # activity = activity_map[filename[5]]

        activity_data[filename] = csi_data

        print("Processing done for file: " + filename + "\n")

        # add activity column
        # csi_data['activity'] = activity

        # df = pd.concat([df, csi_data], ignore_index=True)
```
