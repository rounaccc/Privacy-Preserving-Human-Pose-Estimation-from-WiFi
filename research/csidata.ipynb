```python
import scipy.io as sio
import h5py
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
%matplotlib inline
```

```python
mat_contents = sio.loadmat('./Dataset/CSI dataset for wireless human sensing/dataset/AR-1a/AR1a_W.mat')
print(mat_contents.keys())
```

```python
mat_contents['csi_buff'].shape
```

```python
df = pd.DataFrame.from_dict(mat_contents['csi_buff'])
```

```python
csi_magnitude = np.abs(mat_contents['csi_buff'])

# Create a heatmap (2D color plot) for the magnitude of the CSI data
plt.imshow(csi_magnitude, cmap='viridis', aspect='auto', origin='lower')
plt.colorbar(label='Magnitude (dB)')
plt.title('CSI Data Heatmap')
plt.xlabel('Subcarrier Index')
plt.ylabel('Time Index')
plt.show()
```
